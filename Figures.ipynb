```python
def add_sw_sol_subplot(ax):

    # Constants
    g = 9.8
    Omega=7.292*10**-5
    a=6.371*10**6
    beta = 2*Omega/a
    to_cpd = 86400.0 / (2.0 * np.pi)

    for h in [12.5,25,50]: # equivalent depth
        c  = np.sqrt(g*h)

        # Define the wavenumber array
        k_idx = np.arange(-20, 20.01, 0.01)
        k_phys = k_idx / a  # in 1/m

        # Initialize storage for real and imaginary parts of roots
        omega_real_parts = []
        omega_imag_parts = []

        for k in k_phys:
            A = (c**2) * (k**2) + 3 * beta * c
            B = beta * c**2 * k
            coeffs = [1, 0, -A, -B]

            roots = np.roots(coeffs)

            # Sort by real part (or magnitude) for consistency
            roots = sorted(roots, key=lambda r: r.real)

            # Append real and imaginary parts of roots
            omega_real_parts.append([r.real * to_cpd for r in roots])
            omega_imag_parts.append([r.imag * to_cpd for r in roots])

        # Convert lists of lists to numpy arrays
        omega_real_parts = np.array(omega_real_parts)
        omega_imag_parts = np.array(omega_imag_parts)

        # Plot real parts of omega
        for i in range(omega_real_parts.shape[1]):
             ax.plot(k_idx, omega_real_parts[:, i], color='gray',linestyle='--',linewidth=0.5)

        # Kelvin wave (real only)
        omega_kelvin = k_phys * c
        ax.plot(k_idx, omega_kelvin * to_cpd, color='gray',linestyle='--',linewidth=0.5)

ds=np.load('fig_spectra_freq.npz')

fig,axs=plt.subplots(2,2,layout='constrained',sharex=True,sharey=True,figsize=(8,8))

levels=np.linspace(-8,8,21)

c=axs[0,1].contourf(ds['x_C96'],ds['y_C96'],ds['z_C96'],levels=levels,cmap='RdBu',extend='both')
c=axs[1,1].contourf(ds['x_C1536'],ds['y_C1536'],ds['z_C1536'],levels=levels,cmap='RdBu',extend='both')
c=axs[0,0].contourf(ds['x_T63'],ds['y_T63'],ds['z_T63'],levels=levels,cmap='RdBu',extend='both')
c=axs[1,0].contourf(ds['x_T255'],ds['y_T255'],ds['z_T255'],levels=levels,cmap='RdBu',extend='both')

cax = fig.add_axes([1.02, 0.15, 0.02, 0.7]) 
cb=plt.colorbar(c,cax=cax)
cb.set_label('Covariance of $v$ and $q$ [$10^{6}m^2s^{-2}$]')

add_sw_sol_subplot(axs[0,0])
add_sw_sol_subplot(axs[1,0])
add_sw_sol_subplot(axs[0,1])
add_sw_sol_subplot(axs[1,1])

axs[0,1].text(0.97, 0.97, 'C96', transform=axs[0,1].transAxes, ha='right', va='top',fontweight='bold')
axs[0,0].text(0.97, 0.97, 'T63', transform=axs[0,0].transAxes, ha='right', va='top',fontweight='bold')
axs[1,1].text(0.97, 0.97, 'C1536', transform=axs[1,1].transAxes, ha='right', va='top',fontweight='bold')
axs[1,0].text(0.97, 0.97, 'T255', transform=axs[1,0].transAxes, ha='right', va='top',fontweight='bold')

axs[0,0].set(xlim=(-20,20),ylim=(0,0.5),ylabel='Frequency [day$^{-1}$]')
axs[1,0].set(ylabel='Frequency [day$^{-1}$]')
axs[1,1].set(xlabel='Zonal Wavenumber')

axs[0,1].set_title('AM4-MG2',fontweight='bold')
axs[0,0].set_title('Idealized GCM',fontweight='bold')
axs[1,0].set(xlabel='Zonal Wavenumber')
axs[1,1].set(xlabel='Zonal Wavenumber')

# makeup

axs[0,0].grid(True)
axs[1,0].grid(True)
axs[0,1].grid(True)
axs[1,1].grid(True)

axs[0,0].text(0.03, 0.97, 'a', transform=axs[0,0].transAxes, ha='left', va='top',fontweight='bold')
axs[1,0].text(0.03, 0.97, 'c', transform=axs[1,0].transAxes, ha='left', va='top',fontweight='bold')
axs[0,0].spines['top'].set_visible(False)
axs[0,0].spines['right'].set_visible(False)
axs[1,0].spines['top'].set_visible(False)
axs[1,0].spines['right'].set_visible(False)

axs[0,1].text(0.03, 0.97, 'b', transform=axs[0,1].transAxes, ha='left', va='top',fontweight='bold')
axs[1,1].text(0.03, 0.97, 'd', transform=axs[1,1].transAxes, ha='left', va='top',fontweight='bold')
axs[0,1].spines['top'].set_visible(False)
axs[0,1].spines['right'].set_visible(False)
axs[1,1].spines['top'].set_visible(False)
axs[1,1].spines['right'].set_visible(False)

plt.show()
```
